```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


eth_df = pd.read_csv('../data/ethiopia.csv').replace(-999, pd.NA)
ken_df = pd.read_csv('../data/kenya.csv').replace(-999, pd.NA)
nig_df = pd.read_csv('../data/nigeria.csv').replace(-999, pd.NA)
sud_df = pd.read_csv('../data/sudan.csv').replace(-999, pd.NA)
tan_df = pd.read_csv('../data/tanzania.csv').replace(-999, pd.NA)

print("All 5 countries (Ethiopia, Kenya, Nigeria, Sudan, Tanzania) loaded successfully.")
```

```python
plt.figure(figsize=(12, 6))


sns.lineplot(data=eth_df, x='DOY', y='T2M', label='Ethiopia')
sns.lineplot(data=ken_df, x='DOY', y='T2M', label='Kenya')
sns.lineplot(data=nig_df, x='DOY', y='T2M', label='Nigeria')
sns.lineplot(data=sud_df, x='DOY', y='T2M', label='Sudan')
sns.lineplot(data=tan_df, x='DOY', y='T2M', label='Tanzania')

plt.title('Daily Temperature Comparison Across 5 African Nations')
plt.xlabel('Day of Year (DOY)')
plt.ylabel('Temperature (°C)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, alpha=0.3)
plt.show()
```

Here I am checking the temperature trends for Ethiopia and Kenya to see which country has a more stable climate.
